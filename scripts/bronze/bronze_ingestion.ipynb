## Step 1: Ingesting Raw Data from USGS API

```python
import requests
import json
import time


landing_path = "/Volumes/workspace/earthquake_db/landing_zone/"

def fetch_and_save_earthquakes():
    url = "https://earthquake.usgs.gov/fdsnws/event/1/query?format=geojson"
    try:
        response = requests.get(url,timeout=30)
        
        if response.status_code == 200:
            data = response.json()
            file_name = f"{landing_path}earthquakes_{int(time.time())}.json"
            
            
            dbutils.fs.put(file_name, json.dumps(data), overwrite=True)
            print(f"✅ The file has been successfully saved in: {file_name}")
        else:
            raise Exception(f"Error fetching data: {response.status_code}")
    except Exception as e:
        print(f"Error fetching data: {str(e)}")
        raise e
fetch_and_save_earthquakes()
```
